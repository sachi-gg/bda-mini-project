```python
import pandas as pd
import numpy as np
import random
from collections import defaultdict

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 120)
np.random.seed(42)
random.seed(42)

adjectives = ["Silent","Broken","Golden","Hidden","Last","Crimson","Lost","Midnight","Electric",
              "Frozen","Wild","Secret","Burning","Endless","Shattered","Distant","Iron","Velvet",
              "Neon","Forgotten","Rising","Fallen","Hollow","Bright","Dark"]
nouns = ["River","City","Empire","Dream","Storm","Garden","Kingdom","Horizon","Shadow","Journey",
         "Promise","Mirror","Island","Signal","Legend","Highway","Harbor","Echo","Mountain","Code",
         "Oath","Frontier","Voyage","Whisper","Crown","Circus","Tide","Orbit","Lantern","Season"]

genre_info = [("Drama",0.25,7.0), ("Comedy",0.20,6.2), ("Action",0.18,6.0), ("Thriller",0.12,6.6),
              ("Sci-Fi",0.08,6.8), ("Romance",0.07,6.1), ("Horror",0.06,5.5), ("Animation",0.04,7.3)]

n = 600
titles = random.sample([f"{a} {b}" for a in adjectives for b in nouns], n)
genres = np.random.choice([g[0] for g in genre_info], size=n, p=[g[1] for g in genre_info])
ratings = np.array([np.clip(np.random.normal({g[0]: g[2] for g in genre_info}[g], 0.9), 1.0, 9.8) for g in genres]).round(1)

clean_df = pd.DataFrame({"title": titles, "genre": genres, "rating": ratings})

# Convert to records for MapReduce and save to CSV for R
records = clean_df.to_dict("records")
print(f"Dataset generated with {len(records)} records ready for MapReduce.")
```

```python
def run_mapreduce(records, mapper, reducer):
    # 1. MAP PHASE: Apply mapper to every record
    mapped = []
    for record in records:
        mapped.extend(mapper(record))

    # 2. SHUFFLE / SORT PHASE: Group values by key, then sort keys alphabetically
    grouped = defaultdict(list)
    for key, value in mapped:
        grouped[key].append(value)
    shuffled = dict(sorted(grouped.items()))

    # 3. REDUCE PHASE: Apply reducer to every key group
    reduced = [reducer(key, values) for key, values in shuffled.items()]

    return mapped, shuffled, reduced

def mapper_genre_count(record):
    return [(record["genre"], 1)]

def reducer_sum(key, values):
    return (key, sum(values))
```

```python
sample = records[:10]

print("--- INPUT DATA (10 records) ---")
for r in sample:
    print(f"{r['title']:<22} | {r['genre']}")

mapped, shuffled, reduced = run_mapreduce(sample, mapper_genre_count, reducer_sum)

print("\n--- STEP 1: MAP PHASE (Key, Value) ---")
for k, v in mapped:
    print(f"({k}, {v})")

print("\n--- STEP 2: SHUFFLE & SORT PHASE (Group by Key) ---")
for k, v in shuffled.items():
    print(f"{k:<10} -> {v}")

print("\n--- STEP 3: REDUCE PHASE (Aggregate) ---")
for k, v in reduced:
    print(f"{k:<10} =  {v}")
```

```python
# Execute MapReduce on all 600 records
_, _, final_reduced = run_mapreduce(records, mapper_genre_count, reducer_sum)

# Format final output and save for R visualization
genre_count_df = pd.DataFrame(final_reduced, columns=["genre", "count"]).sort_values("count", ascending=False)
genre_count_df.to_csv("mapreduce_output.csv", index=False)

print("FINAL MAPREDUCE RESULT: Movie Count by Genre")
display(genre_count_df)
```
